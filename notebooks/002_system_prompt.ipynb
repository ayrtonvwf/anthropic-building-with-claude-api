```python
# Create an API client

from anthropic import Anthropic

client = Anthropic()
model = "claude-haiku-4-5"
```

```python
from anthropic.types import MessageParam
from collections.abc import Iterable

def add_user_message(messages: list[MessageParam], text: str):
    user_message: MessageParam = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages: list[MessageParam], text: str):
    assistant_message: MessageParam = {"role": "assistant", "content": text }
    messages.append(assistant_message)

from anthropic.types import TextBlock, Message

def get_message_text(message: Message):
    return next(
        (block.text for block in message.content if isinstance(block, TextBlock)),
        "" # empty string if none
    )

def chat(messages: Iterable[MessageParam], system: str|None = None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }

    if system:
        params["system"] = system

    message = client.messages.create(**params)
    return get_message_text(message)
```

```python
# Make a starting list of messages

messages = []

# Loop forever

while True:
    # Get user input
    user_input = input("> ")

    # Stop if the message is empty
    if not user_input.__len__():
        print("Finishing chat")
        break

    print("> ", user_input)

    # Add user message
    add_user_message(messages, user_input)

    # Define the system prompt

    system = """"
    You are a patient tutor.
    Do not directly answer student's questions.
    Guide them to a solution step by step.
    Use examples close to their problem.
    """

    # Pass the list of messages into 'chat' to get an answer
    answer = chat(messages, system)

    # Output the answer
    print("---")
    print(answer)
    print("---")

    # Take the answer and add it as an assistant message into our list

    add_assistant_message(messages, answer)
```
